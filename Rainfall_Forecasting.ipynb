# Rainfall Time Series Forecasting


# 1. About Rainfall forecasting
## 1. Project Overview

Rainfall Time Series Forecasting is a predictive analytics project that aims to analyse historical rainfall patterns and forecast future rainfall using Time Series Analysis, Statistical Modelling, and Machine Learning techniques.

The project uses historical rainfall observations recorded over time to identify trends, seasonal variations, and other temporal patterns. By learning from these historical observations, forecasting models will be developed to estimate future rainfall values.

The primary objective is to build a reliable forecasting system that can provide meaningful insights into rainfall behaviour and support better planning and decision-making.

## 2. Problem Statement

Rainfall patterns vary across different months and years. Understanding these variations is important for agriculture, water resource management, flood risk assessment, and environmental planning.

Analysing historical rainfall data manually may not be sufficient to identify complex temporal patterns or estimate future rainfall accurately.

Therefore, this project focuses on developing a data-driven forecasting approach that analyses historical rainfall observations and predicts future rainfall based on patterns identified in the data.

## 3. Project Objectives

The main objectives of this project are:

- Analyse historical rainfall observations and understand their characteristics.
- Identify long-term trends and seasonal rainfall patterns.
- Perform data cleaning and preprocessing to prepare the datasets fBBBor analysis.
- Investigate time series properties such as stationarity and autocorrelation.
- Apply suitable statistical and machine learning forecasting techniques.
- Train and compare multiple forecasting models.
- Evaluate model performance using appropriate forecasting metrics.
- Generate future rainfall predictions.
- Visualize historical rainfall patterns and forecasted values.

## 4. Dataset Description

The project includes three datasets containing different rainfall measurements:

| Dataset | Description |
|---|---|
| Total Rainfall | Monthly total rainfall measurements. |
| Maximum Rainfall in a Day | Maximum rainfall recorded on a single day within each month. |
| Number of Rainy Days | Number of rainy days recorded during each month. |

These datasets provide different perspectives on rainfall behaviour and will be explored to understand their structure, time coverage, and suitability for forecasting.

## 5. Methodology

The project will follow a structured time series forecasting workflow:

1. **Data Collection and Understanding:** Load and inspect the provided datasets.
2. **Data Preprocessing:** Handle missing values, duplicates, date formats, and data inconsistencies.
3. **Exploratory Data Analysis (EDA):** Analyse rainfall distributions, trends, seasonal patterns, and relationships.
4. **Time Series Analysis:** Examine stationarity, autocorrelation, and seasonal decomposition.
5. **Feature Engineering:** Create appropriate lag-based and time-based features where required.
6. **Model Development:** Implement suitable statistical and machine learning forecasting models.
7. **Model Evaluation:** Compare forecasting performance using MAE, RMSE, and other appropriate metrics.
8. **Future Forecasting:** Generate predictions for future rainfall periods.
9. **Visualization and Interpretation:** Present actual versus predicted rainfall and interpret the results.

## 6. Forecasting Techniques

The following techniques will be explored based on dataset characteristics and suitability:

**Statistical Models**
- ARIMA (AutoRegressive Integrated Moving Average)
- SARIMA (Seasonal AutoRegressive Integrated Moving Average)
- Exponential Smoothing

**Machine Learning Models**
- Linear Regression with lag features
- Random Forest Regressor
- XGBoost (if suitable)

**Deep Learning Extension**
- LSTM (Long Short-Term Memory)

The models will be compared to identify an appropriate forecasting approach based on their performance on unseen time periods.

## 7. Model Evaluation Metrics

The forecasting models will be evaluated using:

- **MAE (Mean Absolute Error):** Measures the average absolute difference between actual and predicted rainfall.
- **RMSE (Root Mean Squared Error):** Measures prediction errors while giving greater weight to larger errors.
- **MAPE (Mean Absolute Percentage Error):** Measures average percentage prediction error where actual values are non-zero and the metric is appropriate.

## 8. Expected Outcome

The expected outcome is a complete Rainfall Time Series Forecasting System capable of:

#### Detect outliers using the IQR method

```python
columns = [
    "total_rainfall",
    "maximum_rainfall_in_a_day",
    "no_of_rainy_days"
]

for column in columns:
    Q1 = rainfall_df[column].quantile(0.25)
    Q3 = rainfall_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = rainfall_df[
        (rainfall_df[column] < lower_limit) |
        (rainfall_df[column] > upper_limit)
    ]

    print(f"\n{column}")
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))
```

### Stationarity test (ADF test)

```python
from statsmodels.tsa.stattools import adfuller

# Perform ADF test
adf_result = adfuller(rainfall_ts.dropna())

print("ADF Statistic:", adf_result[0])
print("p-value:", adf_result[1])
print("Critical Values:")

for key, value in adf_result[4].items():
    print(f"{key}: {value}")

if adf_result[1] <= 0.05:
    print("\nResult: The series is stationary.")
else:
    print("\nResult: The series is non-stationary.")
```

# 4. Feature Engineering

### Create calendar and cyclical features

```python
# Create a separate dataset for feature engineering

model_df = rainfall_df.copy()

# Extract year and month
model_df["year"] = model_df["month"].dt.year
model_df["month_number"] = model_df["month"].dt.month

# Create cyclical monthly features
model_df["month_sin"] = np.sin(
    2 * np.pi * model_df["month_number"] / 12
)

model_df["month_cos"] = np.cos(
    2 * np.pi * model_df["month_number"] / 12
)

# Display the new features
model_df[
    ["month", "year", "month_number", "month_sin", "month_cos"]
].head(15)
```

### Create lag features

```python
# Create lag features for total rainfall

for lag in [1, 2, 3, 6, 12]:
    model_df[f"rainfall_lag_{lag}"] = (
        model_df["total_rainfall"].shift(lag)
    )

# Create lag features for maximum daily rainfall

for lag in [1, 2, 3, 6, 12]:
    model_df[f"max_daily_lag_{lag}"] = (
        model_df["maximum_rainfall_in_a_day"].shift(lag)
    )

# Display selected columns
model_df[
    [
        "month",
        "total_rainfall",
        "rainfall_lag_1",
        "rainfall_lag_2",
        "rainfall_lag_3",
        "rainfall_lag_6",
        "rainfall_lag_12"
    ]
].head(15)
```

### Create rolling statistical features

```python
# Create rolling features using only previous observations

targets = {
    "total_rainfall": "rainfall",
    "maximum_rainfall_in_a_day": "max_daily"
}

for column, prefix in targets.items():

    previous_values = model_df[column].shift(1)

    for window in [3, 6, 12]:

        model_df[f"{prefix}_rolling_mean_{window}"] = (
            previous_values.rolling(window=window).mean()
        )

    model_df[f"{prefix}_rolling_std_3"] = (
        previous_values.rolling(window=3).std()
    )

# Display the new features
model_df[
    [
        "month",
        "total_rainfall",
        "rainfall_rolling_mean_3",
        "rainfall_rolling_mean_6",
        "rainfall_rolling_mean_12",
        "rainfall_rolling_std_3"
    ]
].head(20)
```
